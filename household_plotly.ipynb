```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
```

```python
df = pd.read_csv("D:/VIS/data_makeovermonday/makeovermonday-2025w39-family-spending-in-the-uk/makeovermonday-2025w39-family-spending-in-the-uk/original/Figure_5__Poorer_households_spent_proportionally_more_on_housing,_fuel_and_power_than_richer_households_in_FYE_2024.csv")
```

```python
df.columns
```

```python
df_long = pd.melt(
    df, 
    id_vars=['% of Total Weekly Expendtiure'], 
    var_name='rank', 
    value_name='percent'
)
```

```python
df_long
```

```python
df_long_filter = df_long.query("rank not in ['All households', '2nd', '3rd', '4th']")
```

```python
df_long_filter.columns
```

```python
fig = px.bar(df_long_filter, 
y= '% of Total Weekly Expendtiure', 
x ='percent', 
color= 'rank', 
template='plotly_white',
color_discrete_sequence= ["#5D74A5", "#A8554E"],
orientation='h', 
barmode= 'group',
width=780, height=460)
fig.update_layout(
    title = dict(
        text = 'Household Spending in The UK: Comparison Between Top and Bottom',
        font=dict(size=24, color="black", family="Roboto", weight='bold'), 
        x=0.5, 
        xanchor='center'), 
yaxis=dict(
       title=dict(
            text="Weekly Spending",
            font=dict(family="Roboto", size=14, color="black")
        )
    ), 
xaxis=dict(
       title=dict(
            text="Percent",
            font=dict(family="Roboto", size=14, color="black")
        )
    ), 
legend_title=dict(
        text="Year",
        font=dict(family="Roboto", size=14, color="black")
    ), 
annotations = [
        dict(
            text = "Data = ONS | Vis by MhKirmizi", 
            xref = "paper", yref= "paper", 
            x = 1, y = -.04, 
            showarrow = False, 
            font = dict(size = 12, color = "black"), 
            align = "left"
        )
    ]
)
fig.show()
fig.write_image("Household_spending.png", width=1920, height=1080, scale=2)
```
