```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-10-21/historic_station_met.csv")
```

```python
df = df[df['year'] > 1889].reset_index().drop('index', axis=1)
```

```python
df_year = df.groupby('year')[['tmax', 'tmin', 'af', 'rain', 'sun']].mean().reset_index()
```

```python
avg_tmax = df_year['tmax'].mean()
avg_tmin = df_year['tmin'].mean()
avg_af = df_year['af'].mean()
avg_rain = df_year['rain'].mean()
avg_sun = df_year['sun'].mean()
```

```python
df_year['tmax_anomaly'] = df_year['tmax'] - avg_tmax 
df_year['tmin_anomaly'] = df_year['tmin'] - avg_tmin 
df_year['af_anomaly'] = df_year['af'] - avg_af
df_year['rain_anomaly'] = df_year['rain'] - avg_rain
df_year['sun_anomaly'] = df_year['sun'] - avg_sun   
```

```python
fig = px.bar(
    data_frame=df_year, 
    x = 'year', 
    y=[1]*len(df_year),
    color = 'tmax_anomaly', 
    color_continuous_scale='RdBu_r'
)
fig.update_layout(
    yaxis = dict(showticklabels =False, showgrid=False, zeroline=False), 
    xaxis = dict(showticklabels =False, showgrid=False, zeroline=False),
    plot_bgcolor = 'white',
    margin=dict(l=20, r=20, t=20, b=20),
    xaxis_title=None,
    yaxis_title=None,
)
fig.update_coloraxes(showscale=False)
fig.update_traces(width=1)
fig.update_xaxes(showline=False, zeroline=False)
fig.write_image("stripes.png", width=1320, height=540, scale=2)
```
