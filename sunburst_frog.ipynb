```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
frogID_data = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-02/frogID_data.csv')
frog_names = pd.read_csv('https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-09-02/frog_names.csv')
```

```python
frog_names_clean = frog_names.dropna(subset = ["scientificName"])
```

```python
frog_names_clean['subfamily'] = frog_names_clean['subfamily'].str.strip()
frog_names_clean['tribe'] = frog_names_clean['tribe'].str.strip()
```

```python
frog_names_clean['subfamily'].unique()
```

```python
frog_names_clean2 = frog_names_clean[frog_names_clean['subfamily'].isin(['Hylid', 'Myobatrachid', 'Microhylidae'])]
```

```python
fig = px.sunburst(
    frog_names_clean2, 
    path = ['subfamily', 'tribe', 'scientificName'],
    color = 'tribe',
    color_discrete_sequence= px.colors.qualitative.Antique,
    width=756, height=756)
fig.update_layout(
    font = dict(
        family = 'Roboto' 
    ), 
    title = dict(
        text = "Family Tree of Australian Frogs", 
        font=dict(size=42, color="#2a475e", family="Montserrat", weight='bold'), 
        x=0.5, 
        xanchor='center'
    ),
    margin=dict(t=90, b=50), 
    annotations = [
        dict(
            text = "Data = FrogID dataset | Vis by MhKirmizi", 
            xref = "paper", yref= "paper", 
            x = .5, y = -.06, 
            showarrow = False, 
            font = dict(size = 14, color = "black"), 
            align = "left"
        )
    ],
    plot_bgcolor="#fbf9f4",
    paper_bgcolor="#fbf9f4"
)
fig.show()
fig.write_html("frog_family_tree.html")
fig.write_image("sunburst_frog.png", width=756, height=756, scale=2)
```
